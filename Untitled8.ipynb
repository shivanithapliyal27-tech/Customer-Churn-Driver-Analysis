```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python

from google.colab import files
import io

uploaded = files.upload()

file_name = next(iter(uploaded))
df = pd.read_csv(io.BytesIO(uploaded[file_name]))

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
```

```python

print("First 5 rows:")
display(df.head())

print("\nDataset dimensions:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()
```

```python

print("Churn counts:")
print(df["Churn"].value_counts())

print("\nChurn percentage:")
print((df["Churn"].value_counts(normalize=True) * 100).round(2))

sns.countplot(data=df, x="Churn")
plt.title("Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")
plt.show()
```

```python

# Check missing values in each column
missing_values = df.isnull().sum()

missing_summary = pd.DataFrame({
    "Column": df.columns,
    "Missing_Values": missing_values.values,
    "Missing_Percentage": (
        df.isnull().sum().values / len(df) * 100
    ).round(2)
})

display(missing_summary[missing_summary["Missing_Values"] > 0])
```

```python

# Check duplicate rows
print("Total duplicate rows:", df.duplicated().sum())

# Check duplicate customer IDs
print("Duplicate customer IDs:", df["customerID"].duplicated().sum())
```

```python

# Convert TotalCharges to numeric
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"].astype(str).str.strip(),
    errors="coerce"
)

# Check missing values after conversion
print("Missing TotalCharges:", df["TotalCharges"].isnull().sum())

# Check tenure for rows with missing TotalCharges
display(df.loc[df["TotalCharges"].isnull(),
               ["customerID", "tenure", "MonthlyCharges", "TotalCharges"]])
```

```python

# Check whether missing TotalCharges belongs to zero-tenure customers
print("Missing TotalCharges with tenure = 0:",
      ((df["TotalCharges"].isnull()) & (df["tenure"] == 0)).sum())

print("Missing TotalCharges with tenure > 0:",
      ((df["TotalCharges"].isnull()) & (df["tenure"] > 0)).sum())
```

```python

# Fill TotalCharges with 0 only for zero-tenure customers
mask = df["TotalCharges"].isnull() & (df["tenure"] == 0)
df.loc[mask, "TotalCharges"] = 0

print("Remaining missing values:", df.isnull().sum().sum())
```

```python

# Calculate total customers and churned customers
total_customers = len(df)
churned_customers = (df["Churn"] == "Yes").sum()
retained_customers = (df["Churn"] == "No").sum()

churn_rate = churned_customers / total_customers * 100

print("Total Customers:", total_customers)
print("Churned Customers:", churned_customers)
print("Retained Customers:", retained_customers)
print("Overall Churn Rate:", round(churn_rate, 2), "%")
```

```python

churn_counts = df["Churn"].value_counts()

plt.figure(figsize=(7, 5))
plt.pie(
    churn_counts,
    labels=churn_counts.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Customer Churn Distribution")
plt.show()
```

```python

contract_churn = pd.crosstab(
    df["Contract"],
    df["Churn"],
    normalize="index"
) * 100

display(contract_churn.round(2))

contract_churn.plot(
    kind="bar",
    stacked=True,
    figsize=(9, 5)
)

plt.title("Churn Distribution by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Percentage of Customers")
plt.xticks(rotation=0)
plt.legend(title="Churn")
plt.tight_layout()
plt.show()
```

```python

# Create tenure groups
df["TenureGroup"] = pd.cut(
    df["tenure"],
    bins=[-1, 12, 24, 48, 72],
    labels=["0-12 Months", "13-24 Months",
            "25-48 Months", "49-72 Months"]
)

tenure_churn = pd.crosstab(
    df["TenureGroup"],
    df["Churn"],
    normalize="index"
) * 100

display(tenure_churn.round(2))

tenure_churn.plot(
    kind="bar",
    stacked=True,
    figsize=(9, 5)
)

plt.title("Churn Distribution by Customer Tenure")
plt.xlabel("Tenure Group")
plt.ylabel("Percentage of Customers")
plt.xticks(rotation=0)
plt.legend(title="Churn")
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Churn",
    y="MonthlyCharges"
)

plt.title("Monthly Charges vs Customer Churn")
plt.xlabel("Churn")
plt.ylabel("Monthly Charges")
plt.tight_layout()
plt.show()
```

```python

# Select categorical features to analyze
categorical_features = [
    "Contract",
    "InternetService",
    "PaymentMethod",
    "TechSupport",
    "OnlineSecurity",
    "PaperlessBilling",
    "SeniorCitizen",
    "Partner",
    "Dependents"
]

# Calculate churn rate for each category
driver_results = []

for feature in categorical_features:
    churn_rates = df.groupby(feature, observed=True)["Churn"].apply(
        lambda x: (x == "Yes").mean() * 100
    )

    for category, rate in churn_rates.items():
        driver_results.append({
            "Feature": feature,
            "Category": category,
            "Churn_Rate": round(rate, 2)
        })

driver_df = pd.DataFrame(driver_results)

display(
    driver_df.sort_values("Churn_Rate", ascending=False)
             .reset_index(drop=True)
)
```

```python

contract_rates = (
    df.groupby("Contract", observed=True)["Churn"]
      .apply(lambda x: (x == "Yes").mean() * 100)
      .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))
sns.barplot(x=contract_rates.index, y=contract_rates.values)

plt.title("Churn Rate by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Churn Rate (%)")
plt.tight_layout()
plt.show()

display(contract_rates.round(2).rename("Churn Rate (%)"))
```

```python

numeric_summary = df.groupby("Churn")[["tenure", "MonthlyCharges", "TotalCharges"]].mean()

display(numeric_summary.round(2))

numeric_summary.T.plot(kind="bar", figsize=(9, 5))

plt.title("Average Customer Metrics by Churn Status")
plt.ylabel("Average Value")
plt.xlabel("Customer Metric")
plt.xticks(rotation=0)
plt.legend(title="Churn")
plt.tight_layout()
plt.show()

```

```python

top_categories = (
    driver_df.sort_values("Churn_Rate", ascending=False)
             .head(10)
             .sort_values("Churn_Rate")
)

plt.figure(figsize=(10, 6))
sns.barplot(
    data=top_categories,
    x="Churn_Rate",
    y="Feature",
    hue="Category",
    dodge=False
)

plt.title("Top 10 Categories by Churn Rate")
plt.xlabel("Churn Rate (%)")
plt.ylabel("Feature")
plt.legend(title="Category", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python

# Export cleaned dataset for Power BI
df.to_csv("Telco_Churn_Cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
print("Rows and columns:", df.shape)
```

```python

# Export churn driver analysis
driver_ranking = (
    driver_df.sort_values("Churn_Rate", ascending=False)
             .reset_index(drop=True)
)

driver_ranking.to_csv("Churn_Driver_Ranking.csv", index=False)

print("Churn driver ranking saved successfully!")
display(driver_ranking.head(10))
```

```python

# Create project KPIs
kpi_df = pd.DataFrame({
    "Metric": [
        "Total Customers",
        "Churned Customers",
        "Retained Customers",
        "Overall Churn Rate (%)",
        "Average Monthly Charges",
        "Average Tenure (Months)"
    ],
    "Value": [
        len(df),
        (df["Churn"] == "Yes").sum(),
        (df["Churn"] == "No").sum(),
        round((df["Churn"] == "Yes").mean() * 100, 2),
        round(df["MonthlyCharges"].mean(), 2),
        round(df["tenure"].mean(), 2)
    ]
})

kpi_df.to_csv("Churn_Project_KPIs.csv", index=False)

display(kpi_df)
```
